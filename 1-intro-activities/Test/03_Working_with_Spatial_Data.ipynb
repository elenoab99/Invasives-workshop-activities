```python
import geopandas as gpd
import matplotlib.pyplot as plt
from pathlib import Path

# CHANGE THIS to your AOI file
AOI_PATH = Path("data/Merged_CleanV3.geojson")

aoi = gpd.read_file(AOI_PATH)
print(aoi.head())
print("CRS:", aoi.crs)
print("Number of features:", len(aoi))
```

```python
# Plot the AOI
ax = aoi.plot(figsize=(8, 8), edgecolor="black")
ax.set_title("Workshop Area of Interest")
plt.show()
```

```python
print(aoi.geometry.geom_type.value_counts())
```

```python
print(aoi.head())
```

```python
aoi.plot(
    figsize=(10, 8),
    markersize=20
)

plt.title("Field Observation Locations")
plt.show()
```
