**Goal:** Understand how Python can automate parts of an invasive-species mapping workflow.

We start with very simple Python. You do not need previous programming experience.

### 🧠 Why are we doing this?

Python lets us turn a series of GIS steps into a repeatable workflow.

```python
print("Hello, Invasive Species Mapping!")
```

```python
# A variable stores information
species = "African Tulip"
year = 2024

print("Species:", species)
print("Year:", year)
```

```python
# Python can perform calculations
area_m2 = 125600
area_ha = area_m2 / 10000

print("Area:", area_ha, "hectares")
```

```python
# A simple decision
if area_ha > 10:
    print("This is a large mapped area.")
else:
    print("This is a small mapped area.")
```
