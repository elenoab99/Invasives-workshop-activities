# 🐍 Activity 2 — Python Basics

**Goal:** Learn the small amount of Python needed for the rest of the workshop.

```python
# Lists allow us to store multiple values
invasive_species = [
    "African Tulip",
    "Cordia alliodora",
    "Merremia peltata"
]

print(invasive_species)
print("Number of species:", len(invasive_species))
```

```python
# Access one item from a list
print(invasive_species[0])
print(invasive_species[1])
```

```python
# A loop repeats an instruction
for species in invasive_species:
    print("Species:", species)
```

```python
# A function lets us reuse a calculation
def square_m_to_hectares(area_m2):
    return area_m2 / 10000

print(square_m_to_hectares(250000))
```

```python
# A dictionary stores related information
project = {
    "site": "Example Site",
    "year": 2024,
    "sensor": "Sentinel-2"
}

print(project["site"])
print(project["sensor"])
```

### 🧠 Why does this matter?

Later we will use:

- **variables** for paths, years and parameters
- **lists** for bands and species
- **loops** to repeat operations
- **functions** to make reusable processing steps
- **dictionaries** to organise project settings
