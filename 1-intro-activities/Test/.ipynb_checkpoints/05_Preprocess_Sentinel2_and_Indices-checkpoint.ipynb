# Activity 5 — Load, cloud-mask and prepare Sentinel-2 features

**Learning objective:** Turn the scenes found in Activity 4 into a clean set of predictor variables for invasive-species classification.

The workflow is:

**Sentinel-2 scenes → cloud mask → reflectance scaling → NDVI → median composite**

```python
from pystac_client import Client
from dask.distributed import Client as DaskClient
from odc.stac import load, configure_s3_access
import geopandas as gpd
import pandas as pd
import numpy as np
import xarray as xr
import folium
import matplotlib.pyplot as plt

import odc.geo.xr  # activates .odc accessor
```

## 1. Repeat the AOI and STAC search

```python
STAC_URL = "https://earth-search.aws.element84.com/v1"
catalog = Client.open(STAC_URL)

aoi_gdf = gpd.read_file("AOI/EfateAOI.geojson")
aoi_wgs84 = aoi_gdf.to_crs("EPSG:4326")
bbox = aoi_wgs84.total_bounds.tolist()

DATE_RANGE = "2025-01-05/2025-11-30"
MAX_CLOUD = 25

items = catalog.search(
    collections=["sentinel-2-c1-l2a"],
    bbox=bbox,
    datetime=DATE_RANGE,
    query={"eo:cloud_cover": {"lt": MAX_CLOUD}},
).item_collection()

print("Scenes found:", len(items))

if len(items) == 0:
    raise ValueError("No Sentinel-2 scenes found. Change the date range or cloud threshold.")
```

## 2. Start Dask and configure cloud access

```python
# Run once per notebook session.
# If you already have a Dask client running, reuse it rather than creating another one.
try:
    dask_client
    print("Using existing Dask client:", dask_client)
except NameError:
    dask_client = DaskClient(
        n_workers=1,
        threads_per_worker=8,
        memory_limit="12GB"
    )
    print(dask_client)

configure_s3_access(cloud_defaults=True, requester_pays=True)
```

## 3. Load the Sentinel-2 bands

We will use the same core variables as the existing workflow:

- Red
- Green
- Blue
- NIR
- SWIR
- SCL (Scene Classification Layer)

```python
data = load(
    items,
    measurements=["red", "green", "blue", "nir08", "swir16", "scl"],
    bbox=bbox,
    chunks={"x": 2048, "y": 2048},
    groupby="solar_day",
)

print(data)
```

## 4. Apply the Sentinel-2 cloud mask

```python
# SCL classes masked in the existing workflow:
# 1 = saturated/defective
# 3 = cloud shadow
# 9 = high-probability cloud
# 10 = thin cirrus

mask_flags = [1, 3, 9, 10]

cloud_mask = ~data.scl.isin(mask_flags)
masked = data.where(cloud_mask)
```

## 5. Scale reflectance and calculate NDVI

```python
# Sentinel-2 C1 L2A reflectance values are scaled by 10,000.
scaled = (masked.where(masked != 0) * 0.0001).clip(0, 1)

# NDVI becomes an additional predictor feature.
scaled["ndvi"] = (
    (scaled.nir08 - scaled.red) /
    (scaled.nir08 + scaled.red)
)

print(scaled)
```

## 6. Create the median composite

```python
median = scaled.median("time").compute()

print("Predictor variables:")
print(list(median.data_vars))
```

```python
median.odc.explore(vmin=0, vmax=0.3)
```

### Key workshop message

NDVI is **not the final invasive-species output**. It is one of the variables supplied to the Random Forest model.

The final model sees a feature vector such as:

`red + green + blue + nir08 + swir16 + ndvi`

for each training point and later for each image pixel.
