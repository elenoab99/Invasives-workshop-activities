# 🗺️ Activity 3 — Working with Spatial Data

**Goal:** Use Python to read and inspect an AOI or other GIS layer.

```python
# Run this only if the packages are not already installed.
# %pip install geopandas matplotlib
```

```python
import geopandas as gpd
import matplotlib.pyplot as plt
from pathlib import Path

# CHANGE THIS to your AOI file
AOI_PATH = Path("data/aoi.gpkg")

aoi = gpd.read_file(AOI_PATH)
print(aoi.head())
print("CRS:", aoi.crs)
print("Number of features:", len(aoi))
```

```python
# Plot the AOI
ax = aoi.plot(figsize=(8, 8), edgecolor="black")
ax.set_title("Workshop Area of Interest")
plt.show()
```

```python
# Reproject to a metric CRS if needed.
# EPSG:32760 is UTM Zone 60S, commonly useful around Fiji.
aoi_utm = aoi.to_crs("EPSG:32760")

print("Area in square metres:", aoi_utm.geometry.area.sum())
print("Area in hectares:", aoi_utm.geometry.area.sum() / 10000)
```
