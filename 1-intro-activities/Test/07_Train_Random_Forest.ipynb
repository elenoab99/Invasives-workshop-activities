```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

## 1. Separate the class labels and predictor variables

```python
# First column = target class
classes = training_array["randomforest"].to_numpy()

# Everything else = Sentinel-2 / NDVI predictor features
feature_columns = [
    col for col in training_array.columns
    if col != "randomforest"
]

observations = training_array[feature_columns].to_numpy()

print("Features:", feature_columns)
print("Observation matrix shape:", observations.shape)
print("Class array shape:", classes.shape)
```

## 2. Create training and validation subsets

The original notebook fits the model to all available training observations and then reports performance on those same observations.

For the workshop, we add a simple hold-out split so participants can see why testing on unseen samples gives a more informative assessment.

```python
X_train, X_test, y_train, y_test = train_test_split(
    observations,
    classes,
    test_size=0.30,
    random_state=42,
    stratify=classes
)

print("Model training samples:", len(X_train))
print("Held-out test samples:", len(X_test))
```

## 3. Train the Random Forest

```python
classifier = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

model = classifier.fit(X_train, y_train)

print("Random Forest trained.")
```

## 4. Preliminary accuracy check

```python
test_predictions = model.predict(X_test)

print("Overall held-out accuracy:", accuracy_score(y_test, test_predictions))
print()
print(classification_report(y_test, test_predictions))
print("Confusion matrix:")
print(confusion_matrix(y_test, test_predictions))
```

## 5. Which features contributed most?

```python
feature_importance = (
    pd.Series(model.feature_importances_, index=feature_columns)
    .sort_values(ascending=False)
)

feature_importance
```
